## 2. Vector Index Inspection (621 Clinical Chunks Across 10 Guidelines)

```python
vectordb = rag_engine.load_index()
print(f"✓ Loaded ChromaDB collection '{rag_engine.COLLECTION_NAME}'")
results = rag_engine.retrieve(vectordb, "fasting plasma glucose criteria for diabetes", k=3)

print("\n--- Sample Retrieval Result ---")
top_doc, score = results[0]
print(f"Score: {score:.4f}")
print(f"Document: {top_doc.metadata.get('document_name')}, Page {top_doc.metadata.get('page_number')}")
print(f"Section: {top_doc.metadata.get('section_title')}")
print(f"Content: \"{top_doc.page_content[:200]}...\"")
```

## 3. Grounded LLM Generation & Citation Enforcement

```python
query = "What is the fasting plasma glucose threshold for diagnosing diabetes?"
answer = rag_engine.generate_grounded_answer(vectordb, query, k=3)

print(f"Question: {query}\n")
print(json.dumps(answer, indent=2, ensure_ascii=False))
```
